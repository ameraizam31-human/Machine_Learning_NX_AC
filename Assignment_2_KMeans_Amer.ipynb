```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
import warnings
warnings.filterwarnings('ignore')

# Create a sample dataset
x, y = make_blobs(
    n_samples = 400,
    centers = 5,
    random_state = 67
)

# Plot The Clusters
plt.figure(figsize=(8,6))
plt.scatter(x[:,0], x[:,1], c=y, cmap='viridis')
plt.title('Amer Data - Clusters Generated Using make_blobs')
plt.xlabel('Feature X')
plt.ylabel('Feature Y')
plt.grid(True)
plt.show()
```

```python
# Elbow method to find optimal K
wcss = []
for k in range(1, 11):
  kmeans = KMeans(n_clusters=k, random_state=21) # TWEAK: Samakan random_state
  kmeans.fit(x)
  wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,6))
plt.plot(range(1, 11), wcss, marker='s', color='red') # TWEAK: Tukar marker bentuk petak ('s') dan garis merah
plt.title('The Elbow Method - Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.grid(True)
plt.show()
```

```python
# Predict New Point
# (Pilih koordinat secara rawak berdasarkan graf kluster )
new_point = [[0.5, -2.5]]
predicted_cluster = kmeans.predict(new_point)
print('Predicted Cluster :', predicted_cluster[0])

# Plot the predicted point
plt.figure(figsize=(8,6))
plt.scatter(x[:,0], x[:,1], c=y, cmap='viridis')
plt.scatter(new_point[0][0], new_point[0][1], c='black', marker='*', s=300) # TWEAK: Star(black)
plt.title('Amer Data - New Point Prediction')
plt.xlabel('Feature X')
plt.ylabel('Feature Y')
plt.grid(True)
plt.show()
```
